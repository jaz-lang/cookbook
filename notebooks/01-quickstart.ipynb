```python
import jaz
from jaz.llm import LiteLLM

# Pick the model all agents use. The model id is a LiteLLM route, 'provider/model'.
# Swap to an Anthropic model (and set ANTHROPIC_API_KEY) if you prefer.
jaz.configure(llm=LiteLLM(model='openai/gpt-5-mini'))
```

## The core idea

A JAZ agent runs a **REPL loop**: the LLM writes Python, executes it, observes the
result, and iterates until it can return a value of the type you asked for with
`ReturnType`.


```python
from jaz import invoke
from jaz.hooks import ReturnType

result = invoke(ReturnType(int), task="Calculate the factorial of 10")
result
```

## Passing inputs

Keyword arguments are made available to the agent's code.

```python
result = invoke(
    ReturnType(list),
    task="Double each number in the list",
    numbers=[1, 2, 3, 4, 5],
)
result
```

## Returning structured data

Ask for a richer `return_type` and the agent will build it.

```python
summary = invoke(
    ReturnType(dict),
    task="Summarize this text into a dict with keys 'topic' and 'sentiment'.",
    text="JAZ makes agents reason in code, which is a joy to debug.",
)
summary
```
