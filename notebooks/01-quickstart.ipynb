## The core idea

A JAZ agent runs a **REPL loop**: the LLM writes Python, executes it, observes the
result, and iterates until it can return a value of the `return_type` you asked for.

```python
result = jaz.invoke("Calculate the factorial of 10", return_type=int)
result
```

## Passing inputs

Keyword arguments are made available to the agent's code.

```python
result = jaz.invoke(
    "Double each number in the list",
    return_type=list,
    numbers=[1, 2, 3, 4, 5],
)
result
```

## Returning structured data

Ask for a richer `return_type` and the agent will build it.

```python
summary = jaz.invoke(
    "Summarize this text into a dict with keys 'topic' and 'sentiment'.",
    return_type=dict,
    text="JAZ makes agents reason in code, which is a joy to debug.",
)
summary
```
